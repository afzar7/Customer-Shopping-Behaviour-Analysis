```python
import pandas as pd
from sqlalchemy import create_engine

# Read CSV
df = pd.read_csv("cleaned_customer_Behaviour.csv")

# Connect to MySQL
engine = create_engine("mysql+mysqlconnector://root:your_password@localhost/customer_behaviour")

# Create table and insert data
df.to_sql(
    name="my_table",
    con=engine,
    if_exists="replace",  # replace, append, or fail
    index=False
)

print("Data imported successfully")
```

```python
query="""
SELECT * FROM customer_behaviour.customer
"""
df=pd.read_sql(query,engine)
print(df.head(4))
```

```python
#Question1- How Many of Customer_Id's are Subscribed
query = """
SELECT Subscription_Status,
Count(Customer_Id) AS Subscribe
FROM customer_behaviour.customer
GROUP BY Subscription_Status
"""

df = pd.read_sql(query, engine)
print(df)
```

```python
#Question2- Which shipping type is preferred most?
query = """
SELECT Shipping_Type,
COUNT(*) AS Preferred_Shipping_Type
FROM customer_behaviour.customer
GROUP BY Shipping_Type
ORDER BY  Preferred_Shipping_Type Desc
"""

ship = pd.read_sql(query, engine)

print(ship)
```

```python
#Question3- Which season had the highest revenue?
query = """
SELECT Season,
SUM(Purchase_Amount) AS Highest_Sales
FROM customer_behaviour.customer
GROUP BY Season
ORDER BY Highest_Sales Desc
"""

sales = pd.read_sql(query, engine)

print(sales)
```

```python
#Question4- Which category generates the highest revenue?
query = """
SELECT Category,
SUM(Purchase_Amount) AS Highest_Revenue
FROM customer_behaviour.customer
GROUP BY Category
ORDER BY Highest_Revenue Desc
"""

revenue = pd.read_sql(query, engine)

print(revenue)
```

```python
#Question5-Which locations generate the highest revenue?
query="""
SELECT Location,
SUM(Purchase_Amount) AS Highest_Revenue_Generating_Locations
FROM customer_behaviour.customer
GROUP BY Location
ORDER BY Highest_Revenue_Generating_Locations desc
LIMIT 5
"""
high_revenue=pd.read_sql(query,engine)
print("Top_5\n",high_revenue)
```

```python
#Question6-Average Revenue Generated from each Gender during a time period
query="""
select Gender,Frequency_of_Purchases,
avg(Purchase_Amount) as Revenue
FROM customer_behaviour.customer
group by Gender,Frequency_of_Purchases
order by Revenue desc
limit 5
"""
revenue=pd.read_sql(query,engine)
print("Top_5\n",revenue)
```

```python
#Question7-What is the revenue generated  by male vs female customers?
query = """
SELECT Gender,
SUM(Purchase_Amount) AS Revenue,'$'
FROM customer_behaviour.customer
GROUP BY Gender
"""

revenue= pd.read_sql(query, engine)
print(revenue)
```

```python
#Question8- Which payment method is most frequently used?
query = """
SELECT Payment_Method,
COUNT(*) AS Frequently_Used_Method
FROM customer_behaviour.customer
GROUP BY Payment_Method
ORDER BY  Frequently_Used_Method Desc
"""

payment = pd.read_sql(query, engine)

print(payment)
```

```python
#Question9-Where are the most customers are from?
query="""
select Location,
count(Customer_Id) as Id_count
FROM customer_behaviour.customer
group by Location
order by Id_count desc
limit 5
"""
Location=pd.read_sql(query,engine)
print(Location)
```

```python
#Question10-Which age group gave an average rating?
query = """
SELECT Gender,Age_Group,
round(Avg(Review_Rating),2) AS Rating
FROM customer_behaviour.customer
GROUP BY Gender,Age_Group
order by Rating desc
"""

Rating = pd.read_sql(query, engine)
print(Rating)
```

```python
#Question11-Which Category is mostly purschased and which is the most preferred size of customers?
query="""
select Size,Category,
count(Item_Purchased) as Count
FROM customer_behaviour.customer
group by Size,Category
order by Count desc
"""
most_preferred=pd.read_sql(query,engine)
print(most_preferred)
```

```python
#Question12-Which is the most purchased item by each Gender?
query="""
select Item_Purchased,Gender,
count(Gender) as Item_Purchased_Most
FROM customer_behaviour.customer
group by Item_Purchased,Gender
order by Item_Purchased_Most desc
limit 5
"""
item_purchased=pd.read_sql(query,engine)
print(item_purchased)
```

```python
#Question13-Which Category got the highest number of reviews?
query="""
select Category,
sum(Review_Rating) as Review_Rating
FROM customer_behaviour.customer
group by Category
order by Review_Rating desc
"""
high_reviews=pd.read_sql(query,engine)
print(high_reviews)
```

```python
#Question14-How many Customers used the discount?
query = """
SELECT Discount_Applied,
count(Customer_Id) AS Customers_Count
FROM customer_behaviour.customer
GROUP BY  Discount_Applied
"""

df = pd.read_sql(query, engine)
print(df)
```

```python
#Question15-Most Customer's favourite color?
query="""
select Color,
count(Customer_Id) as Purchase_count
FROM customer_behaviour.customer
group by Color
order by Purchase_Count desc
limit 5
"""
fav_color=pd.read_sql(query,engine)
print(fav_color)
```

```python
#Question16-Which season gain more revenue and review?
query="""
select Season,
sum(Purchase_Amount) as Purchase,
sum(Review_Rating) as Review
FROM customer_behaviour.customer
group by Season
order by Review desc
"""
high_revenue=pd.read_sql(query,engine)
print(high_revenue)
```
